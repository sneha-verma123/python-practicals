```python
# Import libraries and load the iris dataset
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
iris = sns.load_dataset('iris')
print(iris.head())
```

```python
# Scatter plot
sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

plt.title("Scatter Plot of Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
#Density plot
sns.kdeplot(
    data=iris,
    x='petal_length',
    hue='species'
)

plt.title("Density Plot of Petal Length")
plt.xlabel("Petal Length")
plt.ylabel("Density")
plt.show()
```

```python
#Violin plot
sns.violinplot(
    data=iris,
    x='species',
    y='petal_length'
)

plt.title("Violin Plot of Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()
```

```python
#Heatmap
corr = iris.drop(columns='species').corr()

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap of Iris Dataset")
plt.show()
```

```python
#Join plot
sns.jointplot(
    data=iris,
    x='petal_length',
    y='petal_width',
    hue='species'
)

plt.show()
```

```python
#Pair plot
sns.pairplot(
    iris,
    hue='species'
)

plt.show()
```
