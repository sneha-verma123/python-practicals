```python
#PROBLEM 1 : Create a data frame
import pandas as pd
import numpy as np

data = {
    'Student': ['David', 'Samuel', 'Terry', 'Evan'],
    'Age': [27, 24, 22, 32],
    'Country': ['UK', 'Canada', 'China', 'USA'],
    'Course': ['Python', 'Data Structures',
               'Machine Learning', 'Web Development'],
    'Marks': [85, 72, 89, 76]
}

df = pd.DataFrame(data)
print(df)

```

```python
#PROBLEM 2 :Retreive the marks column and assign it to variables
b = df['Marks']
print(b)

```

```python
#PROBLEM 3 : Reterive country and course columns and assign them to variable
c = df[['Country', 'Course']]
print(c)

```

```python
#PROBLEM 5 : Data filtering and selection
#A) Retreive customers whose age is greater than 30
print(df[df['Age'] > 30])


```

```python
#B) Calculate the total marks  for each country
print(df.groupby('Country')['Marks'].sum())

```

```python
#C) Select only student , country and marks columns
result = df[['Student', 'Country', 'Marks']]
print(result)

```

```python
#Data transformation and aggregation
#A) Add a new column discounted_price
df['Marks_After_10_Percent_Deduction'] = df['Marks'] * 0.90

print(df)

```

```python
#B) Group the data by country and calculate average marks
average_marks = df.groupby('Country')['Marks'].mean()
print(average_marks)

```
